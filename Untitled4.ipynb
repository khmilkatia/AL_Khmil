```python
import gradio as gr

def enter_date(full_name, adress, phone_number, date, receiver_name, time, belongings):

  document = f"""
Прізвище, ім’я, по батькові: {full_name}

Місце проживання (адреса): {adress}

Контактний телефон: {phone_number}

Дата: {date}

                                        ЗАЯВА

                                Про затримку товару з TEMU

Шановний(а) {receiver_name} !

Прошу прийняти заяву про затримку товару з TEMU.

У період {time} {date} повинна була прийти посилка за адресою: {adress}.

Опис замовленого майна:
{belongings}

Заявник проінформований про право на захист особистих даних згідно з вимогами правил використаня програми та доставки “Про захист персональних даних”.

Прошу вжити необхідних заходів для встановлення обставин затримки та урахувати знижку на наступне замовлення згідно з правилами на платформі.

З повагою,

{full_name}

  """

  return document


demo = gr.Interface(
    fn=enter_date,
    inputs=[
        gr.Textbox(label="ПІБ"),
        gr.Textbox(label="Адреса"),
        gr.Textbox(label="Контактний телефон"),
        gr.Textbox(label="Дата"),
        gr.Textbox(label="Ім'я керівника відділу"),
        gr.Textbox(label="Період доставки"),
        gr.Textbox(label="Опис майна")
    ],
    outputs=gr.Textbox(lines=30),
    title="Генератор заяви про затримку доcтавки з TEMU"
)

demo.launch(share=True)

```
